```python
# In a new Jupyter cell
import os
import glob
from datetime import datetime

# Find all checkpoints
ckpt_pattern = "checkpoints/meta_epoch_*.pt"
ckpts = glob.glob(ckpt_pattern)

if ckpts:
    ckpts_sorted = sorted(ckpts, key=os.path.getmtime, reverse=True)
    print("Available checkpoints (newest first):")
    for ckpt in ckpts_sorted:
        mtime = os.path.getmtime(ckpt)
        timestamp = datetime.fromtimestamp(mtime).strftime('%Y-%m-%d %H:%M:%S')
        print(f"  {ckpt} - Modified: {timestamp}")
    print(f"\n✅ Most recent: {ckpts_sorted[0]}")
else:
    print("❌ No checkpoints found!")
```
